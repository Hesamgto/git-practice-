```python
# How to get every unique pair of values in nums 
# This is called a nested loop since we have one loop inside another one (inner and outer loops)

for i in range(len(nums)):
    for j in range(i+1, len(nums)):
        print(nums[i], nums[j])
```

the way this works is j is always bigger than i and not equal to it so it satisgies both constraints we look for.

```python
# Now we want to use everything so far, function + nested loop + return 

def find_pair_sum(nums:list[int], target: int):
    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i]+nums[j]==target:
                return (i, j)
```

```python
find_pair_sum([2, 4, 6, 10], 10)
```
